## 2. The Question: Decision, Action, Cost of a Wrong Call

**The Question:** Which safe content and search signals are most strongly associated with page visibility, clicks, and engagement?

**The Decision:** Content managers and SEO specialists will decide which pages to review, refresh, or protect based on signal analysis and performance trends.

**Who Acts on It:** A content team or SEO manager reviews the ranked queue and decides whether to:
- Refresh/rewrite content
- Improve metadata (title, meta description)
- Protect high-performing pages
- Prune or merge low-performing pages
- Monitor pages for future changes

**Cost of a Wrong Recommendation:**

| Scenario | Consequence |
|----------|-------------|
| **False Positive** (Flagged a page that didn't need work) | Wasted content team time, unnecessary rewrites, potential harm to good pages |
| **False Negative** (Missed a critical page) | Lost traffic, missed opportunities, declining visibility over time |

```python
# CODE CELL: Decision, Action, Cost numbers
print("=" * 60)
print("DECISION, ACTION, COST ANALYSIS")
print("=" * 60)

total_pages = df['content_id'].nunique()
high_impression_pages = len(df[df['impressions_90d'] >= 500])
declining_pages = len(df[df['trend_direction'] == 'down'])

print(f"\n📊 Decision Context:")
print(f"   - Total pages to manage: {total_pages:,}")
print(f"   - High-impression pages (>=500): {high_impression_pages:,}")
print(f"   - Declining pages: {declining_pages:,}")

print("\n💰 Cost of a Wrong Call:")
print("   - False Positive: Wasted team time, unnecessary work")
print("   - False Negative: Lost traffic, missed opportunity")
```
